```python
import jieba

```

```python
seg_list = jieba.cut("百战程序员提供海量高品质课程。", cut_all=True)
print("全模式：" + "/".join(seg_list)) # 全模式，将迭代器转为字符串
```

```python
seg_list = jieba.cut("百战程序员提供海量高品质课程。", cut_all=False)
print("精确模式: " + "/".join(seg_list)) # 精确模式
```

```python
seg_list = jieba.cut("百战程序员提供海量高品质课程。") # 默认是精确模式
print("默认模式: " + "/".join(seg_list))
```
