```python
import jieba.analyse
```

```python
text = "近日，到哈尔滨旅游的游客李小姐（化名）在社交平台发布信息称，她到哈尔滨市著名景点圣·索菲亚教堂旅游时，租了一只白狐合影，结果意外发现白狐没了牙齿。该内容发布后引发网友热议，有网友质疑商贩的行为涉嫌虐待动物。1月6日，上游新闻记者采访了李小姐，她讲述了发布于1月2日没牙白狐图片的来历：去年12月31日下午两点半左右，她到圣·索菲亚教堂游览，并在教堂对面的马路上，碰到了出租白狐拍照的商人。因为白狐憨态可掬，她花了30元的价格租了一只拍照，没想到回家查看照片时，意外发现和她合影的白狐居然没牙齿，于是将此事分享到了社交平台，怀疑商贩是为了挣钱不惜伤害动物。"

seg_list = jieba.cut(text, cut_all=False)
print("分词结果:" + "/".join(seg_list))
```

```python
# 获取关键词
tags = jieba.analyse.extract_tags(text, topK=5)
print("关键字:" + " ".join(tags))

# 获取管检测和权重
tags = jieba.analyse.extract_tags(text, topK=5, withWeight=True)
for word, weight in tags:
    print(word, weight)
```
