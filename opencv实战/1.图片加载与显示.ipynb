```python
import cv2
```

# 加载图像


```python
img = cv2.imread("parrot.png", cv2.IMREAD_COLOR) # 读取图片
# 输出图片类型
print(type(img))
# 输出图像数组
print(img)
# 输出图像形状
print(img.shape)
print(img.dtype)
print(img.size)
```

```python
# 读取图像，缩小为原来的1/2
img = cv2.imread('parrot.png', cv2.IMREAD_REDUCED_COLOR_2)
cv2.imshow('parrot', img)
key = cv2.waitKey(0)

if (key == ord('q')):
    exit()
```
